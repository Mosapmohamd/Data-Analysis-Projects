# What is the distribution of trip durations?

```python
plt.figure(figsize=(10, 5))
sns.histplot(df['duration_sec'], bins=50, kde=True, color="blue")

plt.xlabel("Trip Duration (seconds)")
plt.ylabel("Frequency")
plt.title("Distribution of Trip Durations")

plt.show()
```

# Insights
### Most trips are likely short, with a few long-duration outliers. 
### This helps identify typical ride times and potential anomalies.
------------------------------

# How does trip duration vary by user type? 

```python
plt.figure(figsize=(8, 5))
sns.boxplot(x='user_type', y='duration_sec', data=df)
plt.ylim(0, df['duration_sec'].quantile(0.95))
plt.title("Trip Duration by User Type")
plt.xlabel("User Type")
plt.ylabel("Trip Duration (seconds)")
plt.show()
```

# Relationship between Trip Duration and User Age 

```python
df['age'] = 2025 - df['member_birth_year']  

plt.figure(figsize=(8, 5))
sns.scatterplot(x='age', y='duration_sec', data=df, alpha=0.5)
plt.ylim(0, df['duration_sec'].quantile(0.95))
plt.title("Trip Duration vs. User Age")
plt.xlabel("User Age")
plt.ylabel("Trip Duration (seconds)")
plt.show()
```

# Peak Hours for Bike Usage

```python
df['start_time'] = pd.to_datetime(df['start_time'])
df['hour'] = df['start_time'].dt.hour

plt.figure(figsize=(8, 5))
sns.countplot(x='hour', data=df, color='blue')
plt.title("Peak Hours for Bike Usage")
plt.xlabel("Hour of the Day")
plt.ylabel("Number of Trips")
plt.show()
```

# Bike Usage on Weekdays vs. Weekends 

```python
df['weekday'] = df['start_time'].dt.day_name()

plt.figure(figsize=(8, 5))
sns.countplot(x='weekday', data=df, order=['Monday', 'Tuesday', 'Wednesday', 'Thursday', 'Friday', 'Saturday', 'Sunday'])
plt.title("Bike Usage on Weekdays vs. Weekends")
plt.xlabel("Day of the Week")
plt.ylabel("Number of Trips")
plt.show()
```

# Monthly Bike Usage Trend

```python
df['month'] = df['start_time'].dt.month_name()

plt.figure(figsize=(10, 5))
sns.countplot(x='month', data=df, order=['January', 'February', 'March', 'April', 'May', 'June', 'July', 'August', 'September', 'October', 'November', 'December'])
plt.xticks(rotation=45)
plt.title("Monthly Bike Usage Trend")
plt.xlabel("Month")
plt.ylabel("Number of Trips")
plt.show()
```

# Average Trip Duration by Hour

```python
hourly_duration = df.groupby('hour')['duration_sec'].mean().reset_index()

plt.figure(figsize=(10, 5))
sns.heatmap(hourly_duration.set_index('hour').T, cmap="Blues", annot=True)
plt.title("Average Trip Duration by Hour")
plt.xlabel("Hour of the Day")
plt.ylabel("Trip Duration (seconds)")
plt.show()
```

# Most Popular Start Stations

```python
plt.figure(figsize=(10, 5))
df['start_station_name'].value_counts().nlargest(10).plot(kind='bar', color='blue')
plt.title("Top 10 Start Stations")
plt.xlabel("Station Name")
plt.ylabel("Number of Trips")
plt.xticks(rotation=90)
plt.show()

```

# Most Popular End Stations

```python

plt.figure(figsize=(10, 5))
df['end_station_name'].value_counts().nlargest(10).plot(kind='bar', color='red')
plt.title("Top 10 End Stations")
plt.xlabel("Station Name")
plt.ylabel("Number of Trips")
plt.xticks(rotation=90)
plt.show()
```

# Trip Duration by Start Station

```python
top_stations = df['start_station_name'].value_counts().nlargest(10).index
filtered_df = df[df['start_station_name'].isin(top_stations)]

plt.figure(figsize=(10, 5))
sns.boxplot(x='start_station_name', y='duration_sec', data=filtered_df)
plt.xticks(rotation=90)
plt.ylim(0, df['duration_sec'].quantile(0.95))
plt.title("Trip Duration by Start Station")
plt.xlabel("Start Station")
plt.ylabel("Trip Duration (seconds)")
plt.show()
```

# Stations with Highest Imbalance

```python
station_balance = df['start_station_name'].value_counts() - df['end_station_name'].value_counts()
station_balance.nlargest(10).plot(kind='bar', color='purple')
plt.title("Stations with Highest Start-End Imbalance")
plt.xlabel("Station Name")
plt.ylabel("Imbalance (Starts - Ends)")
plt.xticks(rotation=90)
plt.show()
```

# Most Common Station-to-Station Travel Routes 

```python
df['route'] = df['start_station_name'] + " → " + df['end_station_name']
df['route'].value_counts().nlargest(10).plot(kind='bar', color='orange')
plt.title("Top 10 Station-to-Station Travel Routes")
plt.xlabel("Route")
plt.ylabel("Number of Trips")
plt.xticks(rotation=90)
plt.show()
```

# Age Distribution of Users

```python
df['age'] = 2025 - df['member_birth_year']  

plt.figure(figsize=(8, 5))
sns.histplot(df['age'], bins=30, kde=True, color="blue")
plt.title("Age Distribution of Users")
plt.xlabel("Age")
plt.ylabel("Count")
plt.show()
```

# Bike Usage by Gender

```python
plt.figure(figsize=(8, 5))
sns.countplot(x='member_gender', data=df, palette="coolwarm")
plt.title("Bike Usage by Gender")
plt.xlabel("Gender")
plt.ylabel("Number of Trips")
plt.show()
```

# Proportion of Bike-Sharing Users

```python
df['bike_share_for_all_trip'].value_counts().plot(kind='pie', autopct='%1.1f%%', colors=['blue', 'red'])
plt.title("Proportion of Bike-Sharing Users")
plt.ylabel("")
plt.show()
```

# Bike Usage by Age Groups

```python
age_bins = [10, 20, 30, 40, 50, 60, 70, 80]
df['age_group'] = pd.cut(df['age'], bins=age_bins, labels=["10-20", "21-30", "31-40", "41-50", "51-60", "61-70", "71-80"])

plt.figure(figsize=(8, 5))
sns.countplot(x='age_group', data=df, palette="mako")
plt.title("Bike Usage by Age Group")
plt.xlabel("Age Group")
plt.ylabel("Number of Trips")
plt.show()
```

# Most Used Bikes

```python
df['bike_id'].value_counts().nlargest(10).plot(kind='bar', color='green')
plt.title("Top 10 Most Used Bikes")
plt.xlabel("Bike ID")
plt.ylabel("Number of Trips")
plt.xticks(rotation=90)
plt.show()
```

# Average Trip Duration per Bike

```python
top_bikes = df['bike_id'].value_counts().nlargest(10).index
filtered_df = df[df['bike_id'].isin(top_bikes)]

plt.figure(figsize=(10, 5))
sns.boxplot(x='bike_id', y='duration_sec', data=filtered_df)
plt.xticks(rotation=90)
plt.ylim(0, df['duration_sec'].quantile(0.95))
plt.title("Average Trip Duration per Bike")
plt.xlabel("Bike ID")
plt.ylabel("Trip Duration (seconds)")
plt.show()
```

# Distribution of Bike Usage

```python
plt.figure(figsize=(8, 5))
sns.histplot(df['bike_id'].value_counts(), bins=30, kde=True, color='blue')
plt.title("Distribution of Bike Usage")
plt.xlabel("Number of Trips per Bike")
plt.ylabel("Frequency")
plt.show()
```
